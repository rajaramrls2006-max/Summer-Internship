Basic Visualization Code

```python
from vega_datasets import data
cars = data.cars()

import altair as alt

interval = alt.selection_interval()

chart = alt.Chart(cars).mark_point().encode(
    x='Horsepower',
    y='Miles_per_Gallon',
    color=alt.condition(
        interval,
        'Origin',
        alt.value('lightgray')
    )
).add_params(
    interval
)

chart
```
